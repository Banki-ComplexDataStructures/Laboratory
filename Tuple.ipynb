# Simple tuple assignment examples

# 1. Create a tuple and unpack it into variables.


```python
student = ("Ben", 92, True)
name, mark, has_passed = student

print("Unpacked from tuple:")
print("Name:", name)
print("Mark:", mark)
print("Has passed:", has_passed)
```

# 2. Tuple unpacking directly on the right-hand side (no intermediate variable).


```python
x, y, z = 10, 20, 30
print("\nDirect tuple unpacking:")
print("x:", x, "y:", y, "z:", z)
```

# 3. Swapping two values using tuple unpacking.


```python
a, b = 5, 10
a, b = b, a
print("\nAfter swap:")
print("a:", a, "b:", b)
```

# 4. Function returning a tuple and unpacking the result.


```python
def min_max(numbers):
    """Return the minimum and maximum of a list as a tuple."""
    return min(numbers), max(numbers)

low, high = min_max([3, 7, 2, 9])
print("\nMin and max from function:")
print("Low:", low, "High:", high)
```
